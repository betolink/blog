```python
import numpy as np
import xarray as xr

da = xr.DataArray(
    np.arange(6).reshape(2, 3),
    dims=("y", "x"),
    coords={"y": [10, 20], "x": [1, 2, 3]},
    name="value",
)
print(da)
```

```python
print(f"mean = {da.mean().item():.1f}")
```
